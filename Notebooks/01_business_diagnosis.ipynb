```python
import pandas as pd

df = pd.read_csv("../Data raw/DataCoSupplyChainDataset.csv", encoding="latin1")

print("Shape:", df.shape)
display(df.head())

print("\nColumn names:")
for col in df.columns:
    print(col)

print("\nData types:")
display(df.dtypes)

print("\nMissing values:")
display(df.isna().sum().sort_values(ascending=False).head(20))
```

```python
# Create a clean working copy
clean_df = df.copy()

# Remove unusable or unnecessary personal columns
columns_to_drop = [
    "Product Description",
    "Order Zipcode",
    "Customer Email",
    "Customer Fname",
    "Customer Lname",
    "Customer Password",
    "Customer Street",
    "Product Image"
]

clean_df = clean_df.drop(columns=columns_to_drop)

# Convert dates
clean_df["order date (DateOrders)"] = pd.to_datetime(
    clean_df["order date (DateOrders)"]
)

clean_df["shipping date (DateOrders)"] = pd.to_datetime(
    clean_df["shipping date (DateOrders)"]
)

print("Original shape:", df.shape)
print("Clean shape:", clean_df.shape)
print("Remaining missing values:", clean_df.isna().sum().sum())
```
